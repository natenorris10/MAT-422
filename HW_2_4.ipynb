###2.4.1 - Maximum Likelihood Estimation

Maximum likelihood estimation (MLE) is an effective way to estimate parameters of distributions.  This is done by finding the highest point in a likelihood function, which indicates the maximum likelihood estimate.

The likelihood function is the joint pdf or pf $f(\mathbf{x} | \theta)$ with observations from a random sample, regarded as a function of $θ$, for given values $x_1, x_2, ..., x_n$.

For each observed vector $\mathbf{x}$, we let $δ(\mathbf{x}) \in Ω$ denote a value of $θ \in Ω$, where $Ω$ is the parameter space.  This value of $θ$ indicates the maximum for the likelihood function $f(\mathbf{x}|θ)$.  We let $\hat{θ} = δ(\mathbf{X})$ be the estimator of $\theta$, and thus this estimator is then referred to as the maximum likelihood estimator of $θ$.  Once $\mathbf{X}=\mathbf{x}$ is observed, the value $δ(\mathbf{x})$ is called the maximum likelihood estimate of $θ$.

```python
# Basic example of MLE

def mle(logpdf, data, x0):
  nll = lambda theta: -np.sum(logpdf(data, theta))
  result = minimize(nll, x0, method="BFGS")
  return result

def normal_logpdf(x, theta):
  mu, log_sigma = theta
  return norm.logpdf(x, loc=mu, scale=np.exp(log_sigma))

rng = np.random.default_rng(10)
data = rng.normal(loc=5, scale=2, size=1000)

res = mle(normal_logpdf, data, x0=[0.0, 0.0])
mu_hat, sigma_hat = res.x[0], np.exp(res.x[1])

print(mu_hat, sigma_hat)
print(data.mean(), data.std())
```

###2.4.2 - Linear Regression

Consider n points gathered from an i.i.d. normal distribution.  For some $\mu, \sigma^2$, the probability of the points being picked will define the likelihood function: <br>
$\mathcal{P}(\mu|y) = \prod\limits_{i=1}^n P_Y(y_i|μ,σ^2) = \prod\limits_{i=1}^n \dfrac{1}{σ\sqrt{2π}}e^{-\frac{(y_i - μ)^2}{2σ^2}}$

We understand y to be a random variable, <br>
$y_i = \hat{y} + ϵ$, where $ϵ \sim \mathcal{N}(0, σ^2)$.  Because of this, $y_i$ is a normal variable with a mean that is a linear function of $\mathbf{x}$ and a fixed standard deviation: <br>
$y_i \sim \mathcal{N}(\hat{y}_i, σ^2)$.  Thus, we pick $μ$ in the normal distributions in our product as <br>
$μ_i = \hat{y}_i$.

We can then derive the maximum likelihood estimate

$\hat{β} = \text{arg max }\mathcal{P}(β|y)$ <br>
$= \underset{\beta}{\operatorname*{arg\,max}}\prod\limits_{i=1}^n \dfrac{1}{σ\sqrt{2π}}e^{-\frac{(y_i - \hat{y}_i)^2}{2σ^2}}$ <br>
$= \underset{\beta}{\operatorname*{arg\,max}}\text{ log}(\prod\limits_{i=1}^n \dfrac{1}{σ\sqrt{2π}}e^{-\frac{(y_i - \hat{y}_i)^2}{2σ^2}})$ <br>
$= \underset{\beta}{\operatorname*{arg\,max}} \sum\limits_{i=1}^n \text{log } (\dfrac{1}{σ\sqrt{2π}}) + \text{log } (e^{-\frac{(y_i - \hat{y}_i)^2}{2σ^2}})$ <br>
$= \underset{\beta}{\operatorname*{arg\,max}} \sum\limits_{i=1}^n \text{log } (e^{-\frac{(y_i - \hat{y}_i)^2}{2σ^2}})$ <br>
$= \underset{\beta}{\operatorname*{arg\,max}} \sum\limits_{i=1}^n -\dfrac{(y_i - \hat{y}_i)^2}{2σ^2}$ <br>
$= \underset{\beta}{\operatorname*{arg\,min}} \sum\limits_{i=1}^n (y_i - \hat{y}_i)^2$

###3.2.1 - Limits and Continuity

Let function $f: D \to \mathbb{R}$ be a real-valued function on $D \subseteq \mathbb{R}^d.$  Then $f$ is a limit $L \in \mathbb{R}$ as $\mathbf{x}$ approaches $\mathbf{a}$ if for any $ϵ>0,$ there exists a $δ>0$ such that $|f(\mathbf{x} - L)| < ϵ$ for all $\mathbf{x} \in D ∩ B_δ (\mathbf{a}) \backslash {\mathbf{a}}$.  It is written as: <br>
$\lim\limits_{\mathbf{x} \to \mathbf{a}} f(\mathbf{x}) = L.$

```python
# Various limits

x = sp.symbols("x")

f = (x**2 - 4) / (x - 2)
print(sp.limit(f, x, 2))

print(sp.limit(sp.sin(x) / x, x, 0))

print(sp.limit(1 / x, x, sp.oo))
```

###3.2.2 - Derivatives

Let $f:D \to \mathbb{R}$ where $D ⊆ \mathbb{R}$ and let $x_0 ∈ D$ be an interior point of $D$.  The derivative of $f$ at $x_0$ is: <br>
$f'(x_0) = \dfrac{df(x_0)}{dx} = \lim\limits_{h \to 0} \dfrac{f(x_0 + h) - f(x_0)}{h}$, provided the limit exists.

```python
# Derivative as the slope at particular point
def f_1(x):
  return x**2

h = [1, 0.1, 0.01, 0.001, 0.0001, 0.00001]
for i in h:
  slope = (f_1(4 + i) - f_1(4)) / i
  print(f"Slope of x^2 at x=4 with h={i}: {slope}")
```

```python
# Various derivatives written symbolically

print(sp.diff(x**2, x))
print(sp.diff(sp.sin(x)))
print(sp.diff((x**3/3) + 4*x))
```

Let $f:D \to \mathbb{R}$ where $D \subseteq \mathbb{R}^d$ and let $x_0 \in D$ be an interior point of $D$.  The partial derivative of $f$ at $\mathbf{x}_0$, with respect to $x_i$ is: <br>
$\dfrac{∂f(\mathbf{x}_0)}{∂x_i} = \lim\limits_{h \to 0} \dfrac{f(\mathbf{x}_0 + h\mathbf{e}_i) - f(\mathbf{x}_0)}{h}$, provided the limit exists.

```python
y = sp.symbols("y")
f_xy = (x**2 * y) + (4*y)

# partial derivatives
print(sp.diff(f_xy, x))
print(sp.diff(f_xy, y))

# partial derivatives evaluated at different points
print(sp.diff(f_xy, x).subs({x:6, y:4}))
print(sp.diff(f_xy, y).subs({x:2, y:3}))
```

###3.2.3 - Taylor's Theorem

Taylor's Theorem allows us to approximate a differentiable function around any given point using a polynomial.  It acts as a generalization of the mean value theorem.

Let $f:D \to \mathbb{R}$ where $D ⊆ \mathbb{R}$.  Suppose $f$ has a m times continous derivative on $[a,b]$.  Then <br>
$f(b) = f(a)+(b-a)f'(a) + \dfrac{1}{2} (b-a)^2 f''(a) + \cdots + \dfrac{(b-a)^{m-1}}{(m-1)!} f^{(m-1)}(a) + R_m$, where $R_m = \dfrac{(b-a)^m}{m!} f^{(m)} (a+θ(b-a))$ for some $0 < θ < 1$.

```python
def taylor_poly(f, x, a, n):
  return sum(
      sp.diff(f, x, k).subs(x, a) / sp.factorial(k) * (x - a)**k
      for k in range(n+1)
  )

poly = taylor_poly(sp.sin(x), x, 0, 5)
print(poly)

# Comparing at x=0.5
print(poly.subs(x, 0.5))
print(sp.sin(0.5))
```
